Deciding which seasons are usable

```python
major_leagues = [
    'Premier-league',
    'Laliga',
    'Serie-a',
    'Bundesliga',
    'Ligue-1'
]
```

```python
major_football = football[
    football['League'].isin(major_leagues)
]

print(len(major_football))
```

```python
major_football = major_football.copy()

major_football['usable'] = (
    major_football['Ball_Possession_Home'].notna()
    & major_football['Ball_Possession_Host'].notna()
    & major_football['Goal_Attempts_Home'].notna()
    & major_football['Goal_Attempts_Host'].notna()
)

print(
    major_football.groupby('season_year')['usable']
    .mean()
    .round(3)
)
```

```python
analysis_football = football[
    (football['League'].isin(major_leagues))
    & (football['season_year'] >= '2013/2014')
].copy()

print(len(analysis_football))
```

```python
analysis_football[
    [
        'Ball_Possession_Home',
        'Ball_Possession_Host',
        'Goal_Attempts_Home',
        'Goal_Attempts_Host',
        'home_score',
        'away_score'
    ]
].isna().sum()
```

```python
analysis_football = analysis_football.dropna(
    subset=[
        'Ball_Possession_Home',
        'Ball_Possession_Host',
        'Goal_Attempts_Home',
        'Goal_Attempts_Host',
        'home_score',
        'away_score'
    ]
)

print(len(analysis_football))
```

```python
import numpy as np

analysis_football['home_attacking_efficiency'] = np.where(
    analysis_football['Goal_Attempts_Home'] > 0,
    analysis_football['home_score'] / analysis_football['Goal_Attempts_Home'],
    np.nan
)

analysis_football['away_attacking_efficiency'] = np.where(
    analysis_football['Goal_Attempts_Host'] > 0,
    analysis_football['away_score'] / analysis_football['Goal_Attempts_Host'],
    np.nan
)
```

It seems that these statistics were only accessible around 2013. There was also not enough information during the 24/25 season in Spain, so I removed that. Additionally, there were 18 leagues across 6 countries, but with limited data in the smaller leagues, I decided to take only 5 top flight leagues for this analysis. Matches missing variables required for my analysis were also removed.
