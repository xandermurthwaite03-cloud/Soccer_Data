Before cheking the data, I checked for duplicates, missing values, bad formatting, and impossible values. I found duplicate records respresenting the same match and removed them. I also converted qunatitiatvie columns to numeric formats. 

The reason I chose this data set was because of how vast the informaition spanned. Unfortunately, a large amount of possesion and goal-attempt data was missing in the eariler seasons. It seems that these statistics were only accessible around 2013. There was also not enough information during the 24/25 season in Spain, so I removed that. Additionally, there were 18 leagues across 6 countries, but with limited data in the smaller leagues, I decided to take only 5 top flight leagues for this analysis. Matches missing variables required for my analysis were also removed. 

Finally, I defined new variables for both the home and away teams by dividing goals by total goal attempts. These columns will serve as the driver for my analysis.

```python

# Home team observations
home = analysis_football[[
    'League', 'season_year',
    'Ball_Possession_Home',
    'Goal_Attempts_Home',
    'home_score',
    'home_attacking_efficiency'
]].copy()

home.columns = [
    'league', 'season', 'possession',
    'attempts', 'goals', 'efficiency'
]

# Away team observations
away = analysis_football[[
    'League', 'season_year',
    'Ball_Possession_Host',
    'Goal_Attempts_Host',
    'away_score',
    'away_attacking_efficiency'
]].copy()

away.columns = home.columns

# Combine both sides of every match
team_matches = pd.concat([home, away], ignore_index=True)

# Create a numerical season variable
team_matches['season_start'] = (
    team_matches['season'].str[:4].astype(int)
)

print(team_matches.head())
print("Team observations:", len(team_matches))
```

```python

print(team_matches[
    ['possession', 'attempts', 'goals', 'efficiency']
].describe())

print("Missing efficiency:",
      team_matches['efficiency'].isna().sum())

print("Efficiency greater than 100%:",
      (team_matches['efficiency'] > 1).sum())

print("Negative values:",
      (team_matches['efficiency'] < 0).sum())
```

```python

print(
    team_matches[
        ['possession', 'attempts', 'goals', 'efficiency']
    ].corr().round(3)
)

print("\nPossession-efficiency correlation by league:")

print(
    team_matches.groupby('league').apply(
        lambda x: x['possession'].corr(x['efficiency'])
    ).round(3)
)
```

```python

import statsmodels.formula.api as smf
from sklearn.metrics import mean_absolute_error

# Keep valid observations
model_data = team_matches.dropna(
    subset=['league', 'season_start',
            'possession', 'efficiency']
).copy()

model_data = model_data[
    model_data['efficiency'].between(0, 1)
]

# Train on earlier seasons, test on later seasons
train = model_data[model_data['season_start'] <= 2020]
test = model_data[model_data['season_start'] >= 2021]

# A: League and season
model_a = smf.ols(
    'efficiency ~ C(league) + season_start',
    data=train
).fit()

# B: Add possession
model_b = smf.ols(
    'efficiency ~ C(league) + season_start + possession',
    data=train
).fit()

# C: Different possession relationships by league
model_c = smf.ols(
    'efficiency ~ season_start + possession * C(league)',
    data=train
).fit()

results = []

for name, model in [
    ('A: Baseline', model_a),
    ('B: Possession', model_b),
    ('C: League-adjusted', model_c)
]:
    prediction = model.predict(test).clip(0, 1)

    mae = mean_absolute_error(
        test['efficiency'], prediction
    )

    results.append({
        'Model': name,
        'MAE': mae
    })

results = pd.DataFrame(results)

print(results)
```

```python

baseline_error = results.loc[0, 'MAE']
possession_error = results.loc[1, 'MAE']
adjusted_error = results.loc[2, 'MAE']

possession_improvement = (
    (baseline_error - possession_error)
    / baseline_error * 100
)

league_improvement = (
    (possession_error - adjusted_error)
    / possession_error * 100
)

print(
    f"Adding possession improved MAE by "
    f"{possession_improvement:.2f}%"
)

print(
    f"Adding league-specific possession effects "
    f"improved MAE by {league_improvement:.2f}%"
)
```

```python

# Compare models for predicting goal attempts

attempt_data = team_matches.dropna(
    subset=['league', 'season_start',
            'possession', 'attempts']
).copy()

train_attempts = attempt_data[
    attempt_data['season_start'] <= 2020
]

test_attempts = attempt_data[
    attempt_data['season_start'] >= 2021
]

attempt_a = smf.ols(
    'attempts ~ C(league) + season_start',
    data=train_attempts
).fit()

attempt_b = smf.ols(
    'attempts ~ C(league) + season_start + possession',
    data=train_attempts
).fit()

attempt_c = smf.ols(
    'attempts ~ season_start + possession * C(league)',
    data=train_attempts
).fit()

attempt_results = []

for name, model in [
    ('A: Baseline', attempt_a),
    ('B: Possession', attempt_b),
    ('C: League-adjusted', attempt_c)
]:
    prediction = model.predict(test_attempts)

    mae = mean_absolute_error(
        test_attempts['attempts'], prediction
    )

    attempt_results.append({
        'Model': name,
        'MAE': mae
    })

attempt_results = pd.DataFrame(attempt_results)

print(attempt_results)

baseline = attempt_results.loc[0, 'MAE']
possession = attempt_results.loc[1, 'MAE']
adjusted = attempt_results.loc[2, 'MAE']

print(
    f"\nPossession improvement: "
    f"{(baseline - possession) / baseline * 100:.2f}%"
)

print(
    f"League adjustment improvement: "
    f"{(possession - adjusted) / possession * 100:.2f}%"
)
```

```python

# Compare prediction errors separately for each league

league_results = []

for league in test_attempts['league'].unique():

    league_test = test_attempts[
        test_attempts['league'] == league
    ]

    baseline_predictions = attempt_a.predict(league_test)
    possession_predictions = attempt_b.predict(league_test)

    baseline_mae = mean_absolute_error(
        league_test['attempts'],
        baseline_predictions
    )

    possession_mae = mean_absolute_error(
        league_test['attempts'],
        possession_predictions
    )

    improvement = (
        (baseline_mae - possession_mae)
        / baseline_mae * 100
    )

    league_results.append({
        'League': league,
        'Baseline MAE': baseline_mae,
        'Possession MAE': possession_mae,
        'Improvement (%)': improvement
    })

league_results = pd.DataFrame(league_results)

print(league_results.round(2))
```

```python

# Check improvement in each held-out season

season_results = []

for season in sorted(test_attempts['season'].unique()):

    season_test = test_attempts[
        test_attempts['season'] == season
    ]

    baseline_mae = mean_absolute_error(
        season_test['attempts'],
        attempt_a.predict(season_test)
    )

    possession_mae = mean_absolute_error(
        season_test['attempts'],
        attempt_b.predict(season_test)
    )

    improvement = (
        (baseline_mae - possession_mae)
        / baseline_mae * 100
    )

    season_results.append({
        'Season': season,
        'Baseline MAE': baseline_mae,
        'Possession MAE': possession_mae,
        'Improvement (%)': improvement
    })

season_results = pd.DataFrame(season_results)

print(season_results.round(2))
```
